## WORD2VEC

```python
import re
from gensim.models import Word2Vec

raw_sentences = [
    "Dogs are friendly pets",
    "Cats are playful pets",
    "Dogs and cats need food",
    "A puppy is a young dog",
    "A kitten is a young cat",
    "Dogs enjoy playing with a ball",
    "Cats enjoy playing with toys",
    "Pets need care and attention",
    "The dog eats food",
    "The cat eats food",
]

# Convert sentences to lowercase word lists
sentences = [
    re.findall(r"\b[a-z]+\b", sentence.lower())
    for sentence in raw_sentences
]

# Repeat this small sample corpus so the model gets more training examples
sentences = sentences * 50

model = Word2Vec(
    sentences=sentences,
    vector_size=50,  # Number of values in each word vector
    window=3,        # Look up to 3 words around each target word
    min_count=1,     # Keep words that appear at least once
    sg=1,            # Use Skip-gram , 0 - CBOW , 1 - Skip gram
    workers=1,
    seed=42,
    epochs=100
)

print("Vocabulary size:", len(model.wv))
print("Vector for 'dog':")
print(model.wv["dog"])
```

```python
#Find Similar Words
print(model.wv.most_similar("dog", topn=5))
```

```python
#Comparing 2 words directly
print("dog vs cat:", model.wv.similarity("dog", "cat"))
print("dog vs ball:", model.wv.similarity("dog", "ball"))
```

## GloVe

```python
# Glove
import gensim.downloader as api

# Download and load 50-dimensional GloVe vectors
glove_vectors = api.load("glove-wiki-gigaword-50")

# Get the vector for "coffee"
coffee_vector = glove_vectors["coffee"]

print("Vector length:", len(coffee_vector))
print("First five values:", coffee_vector[:5])

# Find words with vectors similar to "coffee"
similar_words = glove_vectors.most_similar("coffee", topn=5)

for word, score in similar_words:
    print(f"{word}: {score:.3f}")
```

## FastText

```python
from gensim.models import FastText

sentences = [
    ["computer", "science", "uses", "computers"],
    ["computer", "programming", "uses", "software"],
    ["students", "learn", "computer", "science"],
    ["programmers", "write", "computer", "programs"],
    ["computing", "technology", "uses", "computers"],
    ["computer", "systems", "process", "information"],
]

model = FastText(
    sentences=sentences,
    vector_size=50,
    window=3,
    min_count=1,
    min_n=3,
    max_n=6,
    workers=1,
    seed=42,
    epochs=50
)

# Get the vector for "computer"
computer_vector = model.wv["computer"]

print("Vector length:", len(computer_vector))
print("First five values:", computer_vector[:5])

# Find words with vectors similar to "computer"
similar_words = model.wv.most_similar("computer", topn=5)

for word, score in similar_words:
    print(f"{word}: {score:.3f}")
```
