## gensim is a package ND Word2vec is library

## Create a sample training corpus

```python
sentences = [
    ["customer", "purchase", "product", "online"],
    ["business", "sales", "revenue", "profit"],
    ["customer", "service", "support", "feedback"],
    ["product", "price", "quality", "customer"],
    ["marketing", "campaign", "advertisement", "sales"],
    ["sales", "customer", "order", "payment"],
    ["business", "growth", "investment", "profit"],
    ["customer", "feedback", "product", "quality"],
    ["employee", "salary", "business", "management"],
    ["market", "demand", "supply", "product"],
    ["online", "shopping", "customer", "payment"],
    ["sales", "revenue", "business", "growth"]
]

print(f"Loaded {len(sentences)} sample sentences for the business corpus.")
```

## Create the model Word2vec

```python
from gensim.models import Word2Vec
```

```python
model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=1,
    sg=1,
    seed=42,
    epochs=300
)
```

## Display the embedding for a word

```python
vector = model.wv["profit"]
print("Embedding for 'profit':")
print(vector)
print("Vector dimensions:",len(vector))
```

## Compare words using cosine similarity:measures direction similarity between two vectors

```python
similarity = model.wv.similarity("profit","revenue")
print("Similarity:",similarity)
```

## Find similar words

```python
similar_words = model.wv.most_similar("profit",topn=5)
for word,score in similar_words:
  print(word,round(score,3))
```

## Sentence Embedding

## load the libraries

```python
!pip install sentence-transformers
```

## Load a pretrained model

```python
from sentence_transformers import SentenceTransformer
model=SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
```

## Create customer queries

```python
queries=[
   "Where is my order?",
    "How can I track my package?",
    "My order has not been delivered yet.",
    "When will my product arrive?",
    "The delivery status is not updating.",
]

print(f"Loaded {len(queries)} queries.")
```

## Generate sentencce embeddings

```python
embeddings = model.encode(queries)
print("Embedding shape:",embeddings.shape)
```

## Inspect one sentence embedding

```python
print("Original sentence:")
print(queries[0])
print("\nFirst 20 embedding values:")
print(embeddings[0][:20])
```

## Semantic similarity

```python
from sentence_transformers import util
similarity_matrix = util.cos_sim(embeddings,embeddings)
print("Similarity Matrix")
print(similarity_matrix)
```

## Find the most similar customer query

```python
import torch

query_index = 0

scores = similarity_matrix[query_index].clone()

scores[query_index] = -float('inf')

best_match = torch.argmax(scores).item()

print("Customer Query:")
print(sentences[query_index])

print("Most Similar Query:")
print(sentences[best_match])

print("Similarity Score:")
print(scores[best_match].item())
```

## Contextual Embedding

## Install libraries

```python
!pip install transformers torch
```

## Import BERT

```python
import torch
from transformers import AutoTokenizer, AutoModel

model_name = "bert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name)
bert_model = AutoModel.from_pretrained(model_name)
bert_model.eval()
```

## Define the sentences

```python
sentence1 = "I deposited money in the bank."
sentence2 = "We sat beside the river bank."
```

##Tokenize the sentences

```python
inputs1 = tokenizer(sentence1, return_tensors="pt")
inputs2 = tokenizer(sentence2, return_tensors="pt")

print(tokenizer.convert_ids_to_tokens(inputs1["input_ids"][0]))
print(tokenizer.convert_ids_to_tokens(inputs2["input_ids"][0]))
```

## Generate contextual embeddings

```python
with torch.no_grad():
    output1 = bert_model(**inputs1)
    output2 = bert_model(**inputs2)

embedding1 = output1.last_hidden_state
embedding2 = output2.last_hidden_state

print("Sentence 1 shape:", embedding1.shape)
print("Sentence 2 shape:", embedding2.shape)
```

## Extract the representation of "bank"

```python
tokens1 = tokenizer.convert_ids_to_tokens(inputs1["input_ids"][0])
tokens2 = tokenizer.convert_ids_to_tokens(inputs2["input_ids"][0])

bank_index1 = tokens1.index("bank")
bank_index2 = tokens2.index("bank")

bank_vector1 = embedding1[0, bank_index1, :]
bank_vector2 = embedding2[0, bank_index2, :]

print("Bank vector 1 shape:", bank_vector1.shape)
print("Bank vector 2 shape:", bank_vector2.shape)
```

## Compare the contextual embeddings

```python
import torch.nn.functional as F

similarity = F.cosine_similarity(
    bank_vector1.unsqueeze(0),
    bank_vector2.unsqueeze(0)
)
print("Cosine similarity:", similarity.item())
```

## Compare the vectors directly

```python
difference = torch.norm(bank_vector1 - bank_vector2)
print("Euclidean distance:", difference.item())
print("Vectors identical:", torch.allclose(bank_vector1, bank_vector2))
```
